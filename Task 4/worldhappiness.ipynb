NO. (ii) i found no region for 2017 2018 and 2019

we can compare only 2015 and 2016 in these problems

```python
region15 = df15.groupby("Region")["Happiness Score"].mean()
region16 = df16.groupby("Region")["Happiness Score"].mean()
print(region15)
print(region16)
```

```python
regioncompare = pd.DataFrame({
    "2015": region15,
    "2016": region16
})

regioncompare["Average"] = regioncompare.mean(axis=1)

print(regioncompare.sort_values("Average", ascending=False))
```
