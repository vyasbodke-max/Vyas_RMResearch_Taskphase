```python
data = {
    "Name": [
        "Aarav", "Riya", "Kabir", "Ananya", "Vihaan",
        "Ishita", "Arjun", "Meera", "Aditya", "Sara",
        "Rahul", "Nisha", "Karan", "Diya", "Rohan",
        "Tanya", "Dev", "Aanya", "Aryan", "Sneha"
    ],

    "Subject": [
        "Maths", "Physics", "Maths", "Chemistry", "Physics",
        "Maths", "Chemistry", "Physics", "Maths", "Chemistry",
        "Physics", "Maths", "Chemistry", "Physics", "Maths",
        "Chemistry", "Physics", "Maths", "Chemistry", "Physics"
    ],

    "Marks": [
        85, 72, np.nan, 91, 64,
        27, 55, 88, 95, np.nan,
        61, 16, 82, 49, 41,
        34, 51, np.nan, 8, 84
    ],

    "Attendance": [
        92, 85, 78, 95, 88,
        72, 65, 91, 97, 82,
        68, 89, 76, 59, 94,
        83, 71, 62, 87, 90
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.isnull().sum())
```

```python
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
```

```python
print(df.isnull().sum())
```

```python
lowatte = df[df["Attendance"] < 75]

print(lowatte)
```

```python
avgm = df.groupby("Subject")["Marks"].mean()

print(avgm)
```

```python
df["Result"] = np.where(df["Marks"] >= 35, "Pass", "Fail")
df
```

```python
top5 = df.sort_values("Marks", ascending=False).head(5)

print(top5)
```
