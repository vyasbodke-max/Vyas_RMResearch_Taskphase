```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4, 5]
y = [2,4,6,8, 10]
y2 = [1,3,5,7,9]

plt.plot(x, y)
plt.title("Simple line plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.plot(x, y, label="Data 1")
plt.plot(x, y2, label="Data 2")
plt.legend()
plt.show()
```

```python
categories = ["A","B","C","D"]
values = [10,25,15,30]

plt.bar(categories, values)
plt.title("Simple bar chart")
plt.xlabel("Category")
plt.ylabel("Value")
plt.show()
```

```python
marks = [45,50,52,60,62,65,70,72,75,80,85,90]

plt.hist(marks, bins=5)
plt.title("Distribution of marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
x = [1,2,3,4,5]
y = [2,5,4,8,10]

plt.scatter(x, y)
plt.title("Scatter plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
x = [1,2,3,4, 5]

y1 = [2, 4,6, 8,10]
y2 = [1, 3,5,7, 9]

plt.plot(x, y1, label="Line 1")
plt.plot(x, y2, label="Line 2")

plt.title("Multiple Lines")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()
```
